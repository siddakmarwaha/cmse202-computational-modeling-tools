```python
for i in range(len(sonar)):
    if sonar['Class'][i] == 'Rock':
        sonar['Class'][i] = -1
    else:
        sonar['Class'][i] = 1
sonar.head()
```

```python
class Perceptron():

    def __init__ (self, n, labeled_data, iters, learning_rate):        
        # attributes now are: number of features (n), data, weights, iterations, learning rate
        self.data = labeled_data
        self.n=n
        self.weights = np.ones(self.n+1)
        self.iterations = iters
        self.learning_rate = learning_rate
        
    def predict(self, feature_set):
        dot_product = np.dot(feature_set, self.weights) + (self.weights[0])
        if dot_product[0] > 0:
            return 1
        else:
            return -1
        
    def fit(self):
        for i in range(self.iterations):
            for row in range(len(self.data)):
                for k in range(self.n-1):
                    self.weights[k+1] = self.weights[k+1] + (self.learning_rate * (self.data[row][0][2] - self.predict([self.data[row][0]])) * self.data[row][0][k])
                    self.weights[0] = self.weights[0] + (self.learning_rate * (self.data[row][2] - self.predict([self.data[row][0]])))
        return self.weights
        # for all iterations
        #    for each row in the data
        #        find the update value, changes the weights (including bias weight)

    def errors(self):
#         how many rows of the data don't match the provided label?
        errors=0
        for i in range(len(self.data)):
            if self.predict([self.data[i][0],self.data[i][1]]) != self.data[i][2]:
                errors+=1
            else:
                pass
        return errors
    
```

```python
f = open("sonar.csv")
header = next(f) # dump the header line
data = []
for line in f:
    fields = line.split(",")
    # need to strip label because, as the last element, it has a \n
    label = (1.0 if fields[60].strip() == "Mine" else -1.0)
    # the fields are strings until we conver them
    for i in range(60):
        data.append(float(fields[i]))
    data.append(label)
f.close()
p = Perceptron(60, data, 10, 0.01)
p.fit()
p.errors()
# (data)
```
